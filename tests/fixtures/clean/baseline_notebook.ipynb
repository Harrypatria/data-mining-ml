# Logistic baseline separates the two simulated groups
Decision: decide whether the simple model is enough for the pilot.

## Limitations
Simulated data; intervals come from bootstrap resamples of the test set.

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import average_precision_score
from sklearn.model_selection import cross_val_score, train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

SEED = 11
X, y = make_classification(n_samples=1500, weights=[0.9], random_state=SEED)
n_duplicated = len(X) - len(np.unique(X, axis=0))
print(X.shape, int(np.isnan(X).sum()), n_duplicated)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, random_state=SEED)
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
baseline = DummyClassifier(strategy="prior")
for name, est in {"baseline": baseline, "model": model}.items():
    s = cross_val_score(est, X_train, y_train, cv=5, scoring="average_precision")
    print(f"{name}: CV average precision {s.mean():.3f} (sd {s.std():.3f})")
```
